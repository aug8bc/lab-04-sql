```python
type_mapping = {
    "int64": "BIGINT",
    "int32": "INT",
    "float64": "DOUBLE",
    "bool": "TINYINT(1)",
    "datetime64[ns]": "DATETIME",
    "object": "VARCHAR(255)",
    "str": "VARCHAR(255)",
}
```

```python
sql_types = {
    col: type_mapping[str(dtype)]
    for col, dtype in df.dtypes.items()
}
print(sql_types)
```
